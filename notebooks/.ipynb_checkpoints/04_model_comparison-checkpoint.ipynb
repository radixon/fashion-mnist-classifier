## Fetch MLflow Runs

```python
runs = mlflow.search_runs(experiment_names=[mlflow_config['experiment_name']], order_by=["metrics.val_accuracy DESC"], 
                          filter_string="", output_format="pandas")

if runs.empty:
    print("No MLflow runs found. Run 'scripts/train.py' to create MLflow runs.")
else:
    print(f"Found {len(runs)} MLflow runs.")
    # Display Columns
    display(runs[['run_id', 'start_time', 'tags.mlflow.runName', 'params.name', 'metrics.val_accuracy',
                'metrics.val_loss']].head())
```

## Clean and Prepare Data for Analysis

```python
if not runs.empty:
    # Select relevant columns
    comparison_columns = {'run_id':'run_id', 'params.name':'model_name', 'metrics.val_accuracy': 'val_accuracy', 'metrics.val_loss': 'val_loss',
                          'metrics.train_accuracy': 'train_accuracy', 'start_time': 'start_time'}
    comparison_df = runs[comparison_columns.keys()].rename(columns=comparison_columns)

    # Convert metrics
    for col in ['val_accuracy', 'val_loss', 'train_accuracy']:
        comparison_df[col] = pd.to_numeric(comparison_df[col], errors='coerce')

    # Drop rows missing data
    comparison_df.dropna(subset=['model_name', 'val_accuracy'], inplace=True)

    print("Cleaned DataFrame for comparison:")
    display(comparison_df)
```

## Best Run for Each Model

```python
if not runs.empty:
    best_runs_idx = comparison_df.loc[comparison_df.groupby('model_name')['val_accuracy'].idxmax()]
    best_runs_df = best_runs_idx.sort_values('val_accuracy', ascending=False).reset_index(drop=True)

    print("Best performing run for each model architecture:")
    display(best_runs_df)
```

## Visualize Model Comparison

```python
if not runs.empty:
    plt.figure(figsize=(12, 6))

    # Create bar plot
    sns.barplot(data=best_runs_df, x='model_name', y='val_accuracy', palette='viridis')
    plt.title('Model Performance Comparison', fontsize=16)
    plt.xlabel('Model Architecture', fontsize=12)
    plt.ylabel('Best Validation Accuracy', fontsize=12)
    plt.ylim(0.85, 0.95)
    plt.grid(axis='y', linestyle='--', alpha=0.7)

    # Add value labels
    for i, v in enumerate(best_runs_df['val_accuracy']):
        plt.text(i, v + 0.001, f'{v:.4f}', ha='center', va='bottom')

    # Save Plot
    save_path = os.path.join(project_root, 'results', 'figures', 'model_comparison_accuracy.png')
    plt.tight_layout()
    plt.savefig(save_path, dpi=300)
    print(f"Comparison plot saved to: {save_path}")
    plt.show()
```

## Prepare Data for Comparison

```python
models_to_compare = ["VanillaCNN", "DeepCNN", "FashionResNet"]
filtered_runs = runs[runs['params.name'].isin(models_to_compare)].copy()

if filtered_runs.empty:
    print(f"No runs found for models: {models_to_compare}. Check 'params.model.name' in your MLflow logs.")
else:
    best_runs_per_model = filtered_runs.loc[filtered_runs.groupby('params.name')['metrics.val_accuracy'].idxmax()]
    display(best_runs_per_model[['params.name', 'metrics.val_accuracy', 'metrics.val_loss', 'run_id']].reset_index(drop=True))

# Store run IDs
comparison_run_ids = best_runs_per_model['run_id'].tolist()
comparison_model_names = best_runs_per_model['params.name'].tolist()
```

## Plot Learning Curves

```python
if not filtered_runs.empty:
    plt.figure(figsize=(16, 9))

    # Plot Loss
    plt.subplot(1,2,1)
    for model_name in models_to_compare:
        # Get runs for this specific model
        model_runs = filtered_runs[filtered_runs['params.name'] == model_name]
        if not model_runs.empty:
            best_run_id = model_runs.loc[model_runs['metrics.val_accuracy'].idxmax()]['run_id']

            # Fetch metric history
            client = mlflow.tracking.MlflowClient()
            history_loss = client.get_metric_history(best_run_id, "val_loss")
            history_train_loss = client.get_metric_history(best_run_id, "train_loss")
            epochs_loss = [m.step for m in history_loss]
            values_loss = [m.value for m in history_loss]
            epochs_train_loss = [m.step for m in history_train_loss]
            values_train_loss = [m.value for m in history_train_loss]

            # Plot if data exists
            if values_loss:
                plt.plot(epochs_train_loss, values_train_loss, label=f'{model_name} Train Loss', linestyle='--')
                plt.plot(epochs_loss, values_loss, label=f'{model_name} Validation Loss', linestyle='-')
    plt.title('Loss w.r.t. Epochs (Comparison)')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)

    # Plot Accuracy
    plt.subplot(1, 2, 2)
    for model_name in models_to_compare:
        model_runs = filtered_runs[filtered_runs['params.name'] == model_name]
        if not model_runs.empty:
            best_run_id = model_runs.loc[model_runs['metrics.val_accuracy'].idxmax()]['run_id']
            history_accuracy = client.get_metric_history(best_run_id, "val_accuracy")
            history_train_accuracy = client.get_metric_history(best_run_id, "train_accuracy")

            epochs_accuracy = [m.step for m in history_accuracy]
            values_accuracy = [m.value for m in history_accuracy]
            epochs_train_accuracy = [m.step for m in history_train_accuracy]
            values_train_accuracy = [m.value for m in history_train_accuracy]

            # Plot if data exists
            if values_accuracy:
                plt.plot(epochs_train_accuracy, values_train_accuracy, label=f'{model_name} Train Accuracy', linestyle='--')
                plt.plot(epochs_accuracy, values_accuracy, label=f'{model_name} Validation Accuracy', linestyle='-')


    plt.title('Accuracy w.r.t. Epochs (Comaprison)')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.grid(True)

    plt.suptitle("Model Performance Comparison", fontsize=16)
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])

    # Save Plot
    comparison_plot_path = os.path.join(paths_config['figures_save_dir'], 'model_performance_comparison.png')
    plt.savefig(comparison_plot_path)
    plt.show()
    print(f"Model comparison plot saved to: {comparison_plot_path}")
```
